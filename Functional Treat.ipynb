```python
# Function to find factorial
def find_fact(num):
    if num < 0:
        return "Invalid number. Please enter a positive number."
    
    result = 1
    for i in range(1, num + 1):
        result *= i
    return result


# Function to summarize data
def data_summary(data):
    print("\n--- Data Summary ---")
    print("Total Data:", len(data))
    print("Minimum Value:", min(data))
    print("Maximum Value:", max(data))
    print("Sum:", sum(data))
    print("Average:", sum(data) / len(data))


# Function to check even or odd
def check_even_odd(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"


# Function to find prime number
def is_prime(num):
    if num <= 1:
        return False
    
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True


# Main program
def main():
    print("=== Functional Treat Program ===")

    # Factorial
    num = int(input("\nEnter a number to find factorial: "))
    print("Factorial:", find_fact(num))

    # Even/Odd
    print("Number is:", check_even_odd(num))

    # Prime check
    if is_prime(num):
        print("It is a Prime number")
    else:
        print("It is NOT a Prime number")

    # Data summary
    data = [10, 25, 5, 40, 15]
    data_summary(data)


# Run program
main()
```

```python
# Function to calculate total and average
def calculate_result(marks):
    total = sum(marks)
    avg = total / len(marks)
    return total, avg


# Function to find grade
def find_grade(avg):
    if avg >= 90:
        return "A+"
    elif avg >= 75:
        return "A"
    elif avg >= 60:
        return "B"
    elif avg >= 50:
        return "C"
    else:
        return "Fail"


# Function to find highest and lowest marks
def min_max(marks):
    return min(marks), max(marks)


# Function to filter passed students (>= 50)
def filter_pass(marks):
    return list(filter(lambda x: x >= 50, marks))


# Function to square marks (transformation)
def square_marks(marks):
    return list(map(lambda x: x**2, marks))


print("Welcome to Student Marks Analyzer")

while True:
    print("\nMain Menu:")
    print("1. Input Marks")
    print("2. Display Result (Total & Average)")
    print("3. Display Grade")
    print("4. Show Highest & Lowest Marks")
    print("5. Show Passed Marks")
    print("6. Transform Marks (Square)")
    print("7. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        input_marks = input("Enter marks separated by space: ")
        marks = [int(x) for x in input_marks.split()]
        print("Marks added successfully.")

    elif choice == "2":
        total, avg = calculate_result(marks)
        print("Total:", total)
        print("Average:", avg)

    elif choice == "3":
        total, avg = calculate_result(marks)
        print("Grade:", find_grade(avg))

    elif choice == "4":
        mn, mx = min_max(marks)
        print("Minimum Marks:", mn)
        print("Maximum Marks:", mx)

    elif choice == "5":
        passed = filter_pass(marks)
        print("Passed Marks:", passed)

    elif choice == "6":
        transformed = square_marks(marks)
        print("Squared Marks:", transformed)

    elif choice == "7":
        print("Thank you for using the system.")
        break

    else:
        print("Invalid choice. Try again.")
```
